Goal: collect final tables and figures for the report.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
TABLES_DIR = PROJECT_ROOT / "outputs" / "tables"
FIGURES_DIR = PROJECT_ROOT / "outputs" / "figures"
SHAP_DIR = PROJECT_ROOT / "outputs" / "shap_results"

sorted(TABLES_DIR.glob("*.csv")), sorted(FIGURES_DIR.glob("*.png")), sorted(SHAP_DIR.glob("*.png"))
```

```python
baseline_path = TABLES_DIR / "baseline_model_comparison.csv"
threshold_path = TABLES_DIR / "threshold_comparison.csv"
shap_path = TABLES_DIR / "shap_top_features.csv"

tables = {}
for name, path in {
    "baseline_model_comparison": baseline_path,
    "threshold_comparison": threshold_path,
    "shap_top_features": shap_path,
}.items():
    if path.exists():
        tables[name] = pd.read_csv(path)

tables.keys()
```
